```python
import csv

data = [
    ["City", "Region", "Sales", "Quantity"],
    ["Hyderabad", "South", 50000, 2],
    ["Delhi", "North", 15000, 3],
    ["Mumbai", "West", 30000, 4],
    ["Chennai", "South", 20000, 2],
    ["Kolkata", "East", 15000, 5],
    ["Bangalore", "South", 30000, 3]
]

with open("sales_data.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created successfully: sales_data.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("sales_data.csv")

# Scatter plot
plt.scatter(data["Sales"], data["Quantity"])
plt.xlabel("Sales")
plt.ylabel("Quantity")
plt.title("Scatter Plot: Sales vs Quantity")
plt.show()
```
