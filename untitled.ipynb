```r
months <- c("Jan", "Feb", "Mar", "Apr", "May")
sales <- c(100, 120, 150, 130, 170)
plot(sales,
     type = "l",
     xaxt = "n",
     main = "Line Chart",
     xlab = "Months",
     ylab = "Sales")
axis(1, at = 1:5, labels = months)
```

```r
labels <- c("CSE", "IT", "ECE", "EEE")
values <- c(120, 90, 80, 60)

pie(values,
    labels = labels,
    main = "Pie Chart")
```
